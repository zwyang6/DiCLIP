```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11, 1)

DiCLIP = [74.0,76.9,77.4,76.9,78.0,77.8,77.9,77.8,77.7,77.5,77.9]
WeCLIP = [70.6,72.7,73.5,74.4,75.4,74.9,75.0,75.3,75.0,75.2,74.5]
WeakCLIP = [61.6,64.0,68.2,70.5,71.3,72.5,72.3,73.9,73.5,72.6,73.3]
SeCo = [60.0,60.0,69.2,69.6,72.6,72.6,73.8,73.3,74.3,74.1,74.2]

#设置折点属性
xx = ['cornflowerblue','darkcyan','thistle']
c1 = 'indianred'
c2 = 'steelblue'
c3 = 'lightslategray'
c4 = 'orange'
mark = 'H'

plt.rcParams['font.sans-serif'] = 'Arial'
plt.xticks(fontsize=12,fontname='Arial',weight='bold')
x_texts = [str(i*2) + 'k' for i in x]
plt.xticks(x,x_texts)
plt.yticks(fontsize=12,fontname='Arial',weight='bold')
bwith = 1.7 #边框宽度设置为2
TK = plt.gca()#获取边框
TK.spines['bottom'].set_linewidth(bwith)#图框下边
TK.spines['left'].set_linewidth(bwith)#图框左边
TK.spines['top'].set_linewidth(bwith)#图框上边
TK.spines['right'].set_linewidth(bwith)#图框右边

plt.ylim(60,80)
plt.plot(x, DiCLIP, c=c1, ls='-',  marker='p', markersize=7, mfc=c1, mec=c1, mew=1,lw=2.)
plt.plot(x, WeCLIP, c=c3, ls='-',  marker='.', markersize=12, mfc=c3, mec=c3, mew=1,lw=2.)
plt.plot(x, SeCo, c=c4, ls='-', marker=mark, mfc=c4, mec=c4, mew=1,lw=2.,alpha=0.8)
plt.plot(x, WeakCLIP, c=c2, ls='-', marker='s', markersize=5.1, mfc=c2, mec=c2, mew=1,lw=2.)

plt.legend(['DiCLIP (Ours)','WeCLIP','SeCo','WeakCLIP'],fancybox=True,framealpha=1,)
plt.grid('on', ls='-.',lw=1.5,alpha=0.7)
plt.savefig('.//converge_speed2.png',dpi=300)
```
